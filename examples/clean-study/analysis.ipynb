# Survey analysis

Run this notebook from the top after `python clean_data.py`.

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
df = pd.read_csv("data/survey_results.csv")
df.head(3)
```

```python
noise = rng.normal(0, 1, len(df))
df["score_noisy"] = df["score"] + noise
```

```python
df = df[df["age"] > 17]
```

```python
by_region = df.groupby("region")["income"].mean().round(1)
by_region
```
